Crea un nuevo DataFrame con datos de 1 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Considerar lo siguiente a partir de "tutorial-para-la-descarga-de-datos.pdf".

Los datos de las columnas corresponden a:
- Columna A: Año
- Columna B: Mes
- Columna C: Día
- Columna D: Precipitación acumulada
- Columna E: Temperatura máxima
- Columna F: Temperatura mínima

En las celdas donde aparece -99.9 significa que no hay información disponible para esa variable

Luego:
1. Parsear el nombres de la estación Chosica y leer el archivo.
1. Calcula el promedio mensual de precipitación
1. Calcula el promedio anual de precipitación
2. Identifica el día más lluvioso
4. Exporta los resultados formateados en CSV

```python
# Tu código aquí – Ejercicio: Estación

# Leer los datos de las estaciones y crear un DataFrame combinado
import os
files = os.listdir('/content/drive/MyDrive/10MO CICLO/CIENCIA DE DATOS AMBIENTALES /SENAMHI')  # Lista los archivos en el directorio
files
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Parsear el nombre del archivo para obtener el código y nombre de la estación
import re

# Buscamos el archivo de Chosica dentro de la lista
station = [f for f in files if 'chosica' in f.lower()][0]

match = re.search(r'(qc\d+)_(.+)\.txt', station)
codigo = match.group(1)
nombre = match.group(2)

print(f"Código: {codigo}, Nombre: {nombre}")
```

```python
# Leer el archivo CSV de la estación seleccionada
import pandas as pd
station_df = pd.read_table(f'/content/drive/MyDrive/10MO CICLO/CIENCIA DE DATOS AMBIENTALES /SENAMHI/{station}', header=None, sep=r'\s+', na_values=['-999','-99.9']) # Sin encabezados
station_df.columns = ['Año', 'Mes', 'Dia', 'Pr','Tmax', 'Tmin']  # Asignar nombres de columnas
# Formatear las columnas
station_df['Año'] = station_df['Año'].astype(int)
station_df['Mes'] = station_df['Mes'].astype(int)
station_df['Dia'] = station_df['Dia'].astype(int)

station_df.head()
```

```python
# Precipitacion mensual
precipitacion_mensual = station_df.groupby('Mes')['Pr'].mean()

# Precipitacion total anual
precipitacion_anual = station_df.groupby('Año')['Pr'].mean()

# Dia mas lluvioso
dia_mas_lluvioso = station_df.loc[station_df['Pr'].idxmax()]


print(f"Precipitación mensual:\n{precipitacion_mensual}\n")
print(f"Precipitación anual:\n{precipitacion_anual}\n")
print(f"El día más lluvioso fue el {dia_mas_lluvioso['Año']:.0f}-{dia_mas_lluvioso['Mes']:.0f}-{dia_mas_lluvioso['Dia']:.0f} con {dia_mas_lluvioso['Pr']} mm de precipitación.")
```

```python
# Exportar a CSV
resultados_export = pd.DataFrame({
    'Precipitacion_promedio_mensual': precipitacion_mensual
})
resultados_export.to_csv(f'precipitacion_mensual_{nombre}.csv')

precipitacion_anual.to_csv(f'precipitacion_anual_{nombre}.csv')

print("Archivos CSV exportados")
```

### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python
# Tu código aquí – Ejercicio 2: Multiestación

# Parsear el nombre y código de cada estación de la carpeta
import re

estaciones_info = []
for f in files:
    match = re.search(r'(qc\d+)_(.+)\.txt', f)
    codigo = match.group(1)
    nombre = match.group(2)
    estaciones_info.append((f, codigo, nombre))

estaciones_info
```

```python
# Leer cada estación y armar un diccionario {nombre_estacion: serie de precipitacion}
precip_estaciones = {}

for f, codigo, nombre in estaciones_info:
    df = pd.read_table(f'/content/drive/MyDrive/10MO CICLO/CIENCIA DE DATOS AMBIENTALES /SENAMHI/{f}', header=None, sep=r'\s+', na_values=['-999','-99.9'])
    df.columns = ['Año', 'Mes', 'Dia', 'Pr', 'Tmax', 'Tmin']
    df['Fecha'] = pd.to_datetime(dict(year=df['Año'], month=df['Mes'], day=df['Dia']))
    precip_estaciones[nombre] = df.set_index('Fecha')['Pr']

precip_estaciones
```

```python
# Unir estaciones en un solo DataFrame (Fecha, Est1, Est2, ...)
clima_df = pd.DataFrame(precip_estaciones)
clima_df.head()
```

```python
# Promedio mensual de precipitación por estación
promedio_mensual_estaciones = clima_df.groupby(clima_df.index.month).mean()
promedio_mensual_estaciones
```

```python
# Mes más lluvioso x cada estación
mes_mas_lluvioso_estacion = promedio_mensual_estaciones.idxmax()

print(f"Mes más lluvioso por estación:\n{mes_mas_lluvioso_estacion}")
```

```python
# Exportar a CSV
promedio_mensual_estaciones.to_csv('promedio_mensual_estaciones.csv')

print("Archivo CSV exportado")
```
